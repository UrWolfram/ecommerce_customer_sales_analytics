### Q9. How many transactions belong to these non-product categories?

```python
df[df["Description"].str.contains(
    "AMAZON FEE|POSTAGE|Manual|Adjust bad debt|DOTCOM POSTAGE",
    case=False, # case sensitivity
    na=False # NaN values are ignores
    )]["Description"].value_counts()
```

```python
non_products = df[df["Description"].str.contains(
    "AMAZON FEE|POSTAGE|Manual|Adjust bad debt|DOTCOM POSTAGE",
    case=False, # case sensitivity
    na=False # NaN values are ignores
    )]

non_products = pd.DataFrame(non_products)

non_products["Revenue"] = non_products["Quantity"] * non_products["UnitPrice"]

non_products["Revenue"].sum()
```

```python
non_products.groupby(by="Description")["Revenue"].agg(["count", "sum"])
```

### Q10. What are the original transactions of the cancelled ones?

```python
df[df["InvoiceNo"].str.startswith("C")]

df.head()

cancel = df[df["InvoiceNo"] == "C536391"].iloc[0]

matches = df[
    (df["StockCode"] == cancel["StockCode"]) &
    (df["CustomerID"] == cancel["CustomerID"]) &
    (df["Quantity"] == -cancel["Quantity"]) &
    (df["UnitPrice"] == cancel["UnitPrice"]) &
    (df["InvoiceDate"] < cancel["InvoiceDate"])
]

print("Cancelled Transaktion: ")
display(cancel)

print("Original Transaction: ")
display(matches)

```

## Phase 2 – Data Cleaning

We need to sort out, which transaction went through and later got cancelled. These transactions are still in the dataset.

```python
# first we gonna change the CustomerID from a float to an integer

df["CustomerID"] = pd.to_numeric(df["CustomerID"], errors="coerce") # Fehlerhafte Werte werden in NaN umgewandelt

df["CustomerID"] = df["CustomerID"].fillna(0).astype(int)

# remove duplicates

print(len(df))

df = df.drop_duplicates().copy()

print(f"Amount of duplicates: {df.duplicated().sum()}")
print(len(df))

## remove non-product transactions

df_sales = df[~df["Description"].str.contains(
    "AMAZON FEE|POSTAGE|Manual|Adjust bad debt|DOTCOM POSTAGE",
    case=False,
    na=False
)].copy() # dieses sucht eben nach allen non-product transactions und am ende negieren wir das, also nur product transactions

print(len(df_sales))

# remove canceled transactions

df_sales[df_sales["InvoiceNo"].str.startswith("C")]
```
